````python
# ============================================================
# STEP 2.4 — Inspect labels and dataset examples
# ============================================================

def label_to_name(dataset, label_value):
    """
    Convert Hugging Face ClassLabel integer to readable label name.
    Falls back to string conversion if the feature is not ClassLabel.
    """
    try:
        return dataset.features["label"].int2str(int(label_value))
    except Exception:
        return str(label_value)

rows = []

for i in range(len(ds)):
    item = ds[i]
    label_value = item.get("label", None)
    label_name = label_to_name(ds, label_value)

    pdf_obj = item.get("pdf", None)
    pdf_type = type(pdf_obj).__name__

    original_path = None
    if isinstance(pdf_obj, dict):
        original_path = pdf_obj.get("path")
    elif hasattr(pdf_obj, "filename"):
        original_path = getattr(pdf_obj, "filename")
    elif isinstance(pdf_obj, str):
        original_path = pdf_obj

    rows.append({
        "row_idx": i,
        "label_value": label_value,
        "label_name": label_name,
        "pdf_object_type": pdf_type,
        "original_path": original_path,
    })

inspect_df = pd.DataFrame(rows)

display(inspect_df.head(20))
print()
print("Label counts:")
display(inspect_df["label_name"].value_counts().to_frame("count"))
````

````python
# ============================================================
# STEP 2.5 — Save raw Hugging Face files locally
# ============================================================

from PIL import Image
import io
import re
import shutil
from pathlib import Path
import pdfplumber

def safe_name(text):
    """
    Convert text into a filesystem-safe lowercase-ish name.
    """
    text = str(text)
    text = text.strip()
    text = re.sub(r"[^\w\-.]+", "_", text)
    text = re.sub(r"_+", "_", text)
    return text.strip("_") or "unknown"


def detect_extension_from_bytes(data: bytes, fallback=".bin"):
    """
    Guess file extension from magic bytes.
    """
    if data.startswith(b"%PDF"):
        return ".pdf"
    if data.startswith(b"\x89PNG"):
        return ".png"
    if data.startswith(b"\xff\xd8"):
        return ".jpg"
    if data.startswith(b"GIF"):
        return ".gif"
    return fallback


def save_hf_file_object(obj, out_base: Path):
    """
    Save a Hugging Face dataset file object.

    Handles common cases:
    - dict with {'bytes': ..., 'path': ...}
    - PIL Image object
    - local path string
    - raw bytes
    - pdfplumber.pdf.PDF object
    """
    original_path = None

    # Case 1: datasets often returns file objects as dicts.
    if isinstance(obj, dict):
        original_path = obj.get("path")
        data = obj.get("bytes")

        if data is not None:
            suffix = Path(original_path).suffix if original_path else detect_extension_from_bytes(data)
            if not suffix:
                suffix = detect_extension_from_bytes(data)
            out_path = out_base.with_suffix(suffix)
            out_path.write_bytes(data)
            return out_path, original_path

        if original_path:
            src = Path(original_path)
            suffix = src.suffix or ".bin"
            out_path = out_base.with_suffix(suffix)
            shutil.copyfile(src, out_path)
            return out_path, original_path

    # Case 2: PIL Image object.
    if isinstance(obj, Image.Image):
        out_path = out_base.with_suffix(".png")
        obj.save(out_path)
        return out_path, original_path

    # Case 3: raw bytes.
    if isinstance(obj, (bytes, bytearray)):
        suffix = detect_extension_from_bytes(bytes(obj))
        out_path = out_base.with_suffix(suffix)
        out_path.write_bytes(bytes(obj))
        return out_path, original_path

    # Case 4: local file path.
    if isinstance(obj, str):
        src = Path(obj)
        original_path = obj
        suffix = src.suffix or ".bin"
        out_path = out_base.with_suffix(suffix)
        shutil.copyfile(src, out_path)
        return out_path, original_path

    # NEW Case 5: pdfplumber.pdf.PDF object
    if isinstance(obj, pdfplumber.pdf.PDF):
        try:
            # Access the underlying stream
            stream = obj.stream
            if hasattr(stream, 'seek') and callable(stream.seek):
                stream.seek(0)
            pdf_bytes = stream.read()

            out_path = out_base.with_suffix(".pdf")
            out_path.write_bytes(pdf_bytes)
            return out_path, original_path
        except Exception as e:
            raise RuntimeError(f"Failed to extract bytes from pdfplumber.pdf.PDF object: {e}") from e

    raise TypeError(f"Unsupported pdf object type: {type(obj)}")


raw_rows = []

for i in range(len(ds)):
    item = ds[i]

    label_value = item.get("label", None)
    label_name = label_to_name(ds, label_value)
    label_folder = safe_name(label_name)

    pdf_obj = item.get("pdf", None)

    # Use row index first, because some HF objects may not expose stable filenames.
    out_base = RAW_DIR / label_folder / f"row_{i:04d}"
    out_base.parent.mkdir(parents=True, exist_ok=True)

    raw_path, original_path = save_hf_file_object(pdf_obj, out_base)

    raw_rows.append({
        "row_idx": i,
        "label_value": label_value,
        "label_name": label_name,
        "label_folder": label_folder,
        "raw_path": str(raw_path),
        "original_path": original_path,
        "raw_suffix": raw_path.suffix.lower(),
    })

raw_df = pd.DataFrame(raw_rows)
raw_info_path = METADATA_DIR / "raw_files.csv"
raw_df.to_csv(raw_info_path, index=False)

print("Saved raw files metadata to:", raw_info_path)
display(raw_df.head(20))
print()
display(raw_df["label_name"].value_counts().to_frame("count"))
````

````python
# ============================================================
# STEP 2.6 — Define PDF/image rendering and text extraction
# ============================================================

import fitz  # PyMuPDF
import pypdfium2 as pdfium
from PIL import Image, ImageOps

def render_pdf_to_single_png(pdf_path: Path, out_png_path: Path, scale=2.0, gap_px=40):
    """
    Render every page of a PDF and combine pages vertically into one PNG.

    Why combine pages?
    - Most model adapters in this project expect one image path per record.
    - If a PDF contains multiple pages, this avoids silently dropping pages.
    """
    pdf = pdfium.PdfDocument(str(pdf_path))
    rendered_pages = []

    for page_index in range(len(pdf)):
        page = pdf[page_index]
        bitmap = page.render(scale=scale)
        pil_image = bitmap.to_pil().convert("RGB")
        rendered_pages.append(pil_image)

    if len(rendered_pages) == 0:
        raise ValueError(f"No pages found in PDF: {pdf_path}")

    if len(rendered_pages) == 1:
        rendered_pages[0].save(out_png_path)
        return out_png_path

    max_width = max(img.width for img in rendered_pages)
    total_height = sum(img.height for img in rendered_pages) + gap_px * (len(rendered_pages) - 1)

    combined = Image.new("RGB", (max_width, total_height), "white")

    y = 0
    for img in rendered_pages:
        x = (max_width - img.width) // 2
        combined.paste(img, (x, y))
        y += img.height + gap_px

    combined.save(out_png_path)
    return out_png_path


def render_document_to_png(input_path: Path, out_png_path: Path):
    """
    Convert PDF or image file to one RGB PNG.
    """
    input_path = Path(input_path)
    out_png_path = Path(out_png_path)
    out_png_path.parent.mkdir(parents=True, exist_ok=True)

    suffix = input_path.suffix.lower()

    if suffix == ".pdf":
        return render_pdf_to_single_png(input_path, out_png_path)

    if suffix in [".png", ".jpg", ".jpeg", ".webp", ".bmp", ".tif", ".tiff"]:
        img = Image.open(input_path).convert("RGB")
        img.save(out_png_path)
        return out_png_path

    raise ValueError(f"Unsupported document format: {input_path}")


def extract_text_from_pdf(pdf_path: Path):
    """
    Extract embedded text from a PDF using PyMuPDF.
    Works well when the PDF contains selectable text.
    """
    doc = fitz.open(str(pdf_path))
    parts = []

    for page in doc:
        txt = page.get_text("text")
        if txt:
            parts.append(txt)

    return "\n".join(parts).strip()


def extract_text_from_image_with_tesseract(image_path: Path):
    """
    OCR fallback for digital ground-truth documents that are images
    rather than embedded-text PDFs.

    This is used only for typed ground-truth documents, not for evaluating
    handwritten model performance.
    """
    img = Image.open(image_path).convert("RGB")
    text = pytesseract.image_to_string(img)
    return text.strip()


def normalize_ground_truth_text(text: str):
    """
    Light cleanup for ground truth.

    Important:
    - We do not lowercase.
    - We do not remove punctuation.
    - We do not expand abbreviations.
    - We preserve clinical information.
    """
    if text is None:
        return ""

    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # Remove excessive trailing spaces per line.
    lines = [line.rstrip() for line in text.split("\n")]

    # Collapse very large blank gaps, but preserve paragraph/line structure.
    cleaned_lines = []
    blank_count = 0

    for line in lines:
        if line.strip() == "":
            blank_count += 1
            if blank_count <= 1:
                cleaned_lines.append("")
        else:
            blank_count = 0
            cleaned_lines.append(line)

    return "\n".join(cleaned_lines).strip()
````

````python
# ============================================================
# STEP 2.7 — Define pairing-key logic
# ============================================================

def get_source_stem(row):
    """
    Prefer the original HF filename when available.
    Fall back to the saved raw filename.
    """
    original_path = row.get("original_path")
    if isinstance(original_path, str) and original_path.strip():
        return Path(original_path).stem

    raw_path = row.get("raw_path")
    return Path(raw_path).stem


def normalize_pairing_key(stem: str):
    """
    Build a pairing key so the handwritten record and its digital truth
    can be matched.

    Handles likely naming patterns mentioned by the dataset card:
    - 1-001-syn vs 1-001-truth
    - SUNN-PATIENT-001 vs TUNN-PATIENT-001

    The goal is not to be clever; it is to produce a transparent key
    and save unpaired records for inspection.
    """
    s = str(stem).lower()
    s = s.replace(" ", "-").replace("_", "-")
    s = re.sub(r"-+", "-", s)

    # Patient-based naming:
    # SUNN-PATIENT-001 = synthetic
    # TUNN-PATIENT-001 = truth
    #
    # Remove leading S/T before the institution code if this pattern exists.
    m = re.search(r"\b([st])([a-z]+)-patient-0*([0-9]+)\b", s)
    if m:
        institution_core = m.group(2)
        patient_num = int(m.group(3))
        return f"{institution_core}-patient-{patient_num:03d}"

    # Volunteer-based naming:
    # 1-001-syn
    # 1-001-truth
    s = re.sub(r"-(syn|synthetic|handwritten|handwriting|written)$", "", s)
    s = re.sub(r"-(truth|groundtruth|ground-truth|digital|typed)$", "", s)

    # More general cleanup.
    remove_tokens = [
        "synthetic",
        "syn",
        "handwritten",
        "handwriting",
        "medical",
        "document",
        "documents",
        "digital",
        "ground",
        "truth",
        "groundtruth",
        "typed",
        "typewritten",
    ]

    parts = [p for p in re.split(r"[-\s]+", s) if p and p not in remove_tokens]
    key = "-".join(parts)
    key = re.sub(r"-+", "-", key).strip("-")

    return key or s


def is_ground_truth_label(label_name: str):
    label = str(label_name).lower()
    return "truth" in label or "digital" in label or "typed" in label


def is_handwritten_label(label_name: str):
    label = str(label_name).lower()
    return "handwritten" in label or "synthetic" in label or "medical document" in label
````

````python
# ============================================================
# STEP 2.8 — Render documents and extract ground-truth text
# ============================================================

processed_rows = []

for _, row in raw_df.iterrows():
    row_idx = int(row["row_idx"])
    label_name = row["label_name"]
    raw_path = Path(row["raw_path"])
    stem = get_source_stem(row)
    pairing_key = normalize_pairing_key(stem)

    rendered_path = IMAGE_DIR / safe_name(label_name) / f"row_{row_idx:04d}.png"
    rendered_path.parent.mkdir(parents=True, exist_ok=True)

    # Render PDF/image into a clean PNG for model input and visual inspection.
    try:
        render_document_to_png(raw_path, rendered_path)
        render_status = "ok"
        render_error = ""
    except Exception as e:
        render_status = "failed"
        render_error = repr(e)
        rendered_path = None

    gt_text_path = ""
    gt_text = ""
    gt_extraction_method = ""

    # Extract text only for digital ground truth rows.
    if is_ground_truth_label(label_name):
        try:
            if raw_path.suffix.lower() == ".pdf":
                embedded_text = extract_text_from_pdf(raw_path)
                embedded_text = normalize_ground_truth_text(embedded_text)

                if embedded_text:
                    gt_text = embedded_text
                    gt_extraction_method = "pdf_embedded_text"
                else:
                    # Fallback: OCR the rendered typed/digital page.
                    gt_text = extract_text_from_image_with_tesseract(rendered_path)
                    gt_text = normalize_ground_truth_text(gt_text)
                    gt_extraction_method = "tesseract_on_rendered_image"
            else:
                gt_text = extract_text_from_image_with_tesseract(rendered_path)
                gt_text = normalize_ground_truth_text(gt_text)
                gt_extraction_method = "tesseract_on_image"

            gt_out = GT_DIR / f"{pairing_key}.txt"
            gt_out.write_text(gt_text, encoding="utf-8")
            gt_text_path = str(gt_out)

        except Exception as e:
            gt_text = ""
            gt_text_path = ""
            gt_extraction_method = f"failed: {repr(e)}"

    processed_rows.append({
        "row_idx": row_idx,
        "label_value": row["label_value"],
        "label_name": label_name,
        "raw_path": str(raw_path),
        "rendered_image_path": str(rendered_path) if rendered_path else "",
        "original_path": row.get("original_path"),
        "source_stem": stem,
        "pairing_key": pairing_key,
        "is_ground_truth": is_ground_truth_label(label_name),
        "is_handwritten": is_handwritten_label(label_name),
        "render_status": render_status,
        "render_error": render_error,
        "ground_truth_text_path": gt_text_path,
        "ground_truth_text_length": len(gt_text),
        "ground_truth_extraction_method": gt_extraction_method,
    })

processed_df = pd.DataFrame(processed_rows)
processed_path = METADATA_DIR / "processed_files.csv"
processed_df.to_csv(processed_path, index=False)

print("Saved processed file metadata to:", processed_path)
display(processed_df.head(20))

print()
print("Render status counts:")
display(processed_df["render_status"].value_counts().to_frame("count"))

print()
print("Document type counts:")
display(processed_df[["is_ground_truth", "is_handwritten"]].value_counts().to_frame("count"))
````

````python
# ============================================================
# STEP 2.9 — Pair handwritten documents with digital ground truth
# ============================================================

truth_df = processed_df[processed_df["is_ground_truth"] == True].copy()
hand_df = processed_df[processed_df["is_handwritten"] == True].copy()

print("Ground-truth documents:", len(truth_df))
print("Handwritten documents:", len(hand_df))

manifest_rows = []
used_truth_indices = set()
used_hand_indices = set()

# First pass: exact pairing_key matches.
for hand_idx, hand_row in hand_df.iterrows():
    key = hand_row["pairing_key"]
    matches = truth_df[truth_df["pairing_key"] == key]

    if len(matches) == 1:
        truth_idx = matches.index[0]
        truth_row = matches.iloc[0]

        manifest_rows.append({
            "record_id": "",  # filled later
            "pairing_key": key,
            "pair_method": "exact_pairing_key",
            "pair_confidence": "high",

            "handwritten_row_idx": int(hand_row["row_idx"]),
            "handwritten_raw_path": hand_row["raw_path"],
            "handwritten_image_path": hand_row["rendered_image_path"],
            "handwritten_original_path": hand_row["original_path"],
            "handwritten_source_stem": hand_row["source_stem"],

            "ground_truth_row_idx": int(truth_row["row_idx"]),
            "ground_truth_raw_path": truth_row["raw_path"],
            "ground_truth_image_path": truth_row["rendered_image_path"],
            "ground_truth_original_path": truth_row["original_path"],
            "ground_truth_source_stem": truth_row["source_stem"],
            "ground_truth_text_path": truth_row["ground_truth_text_path"],
        })

        used_hand_indices.add(hand_idx)
        used_truth_indices.add(truth_idx)

# Second pass: fallback sorted pairing for remaining rows.
# This is useful if HF filenames are hidden and all rows are only row_XXXX.
remaining_hand = hand_df.drop(index=list(used_hand_indices), errors="ignore").copy()
remaining_truth = truth_df.drop(index=list(used_truth_indices), errors="ignore").copy()

remaining_hand = remaining_hand.sort_values(["source_stem", "row_idx"])
remaining_truth = remaining_truth.sort_values(["source_stem", "row_idx"])

fallback_n = min(len(remaining_hand), len(remaining_truth))

for j in range(fallback_n):
    hand_row = remaining_hand.iloc[j]
    truth_row = remaining_truth.iloc[j]

    manifest_rows.append({
        "record_id": "",
        "pairing_key": f"fallback_sorted_{j+1:04d}",
        "pair_method": "fallback_sorted_within_label",
        "pair_confidence": "low_check_manually",

        "handwritten_row_idx": int(hand_row["row_idx"]),
        "handwritten_raw_path": hand_row["raw_path"],
        "handwritten_image_path": hand_row["rendered_image_path"],
        "handwritten_original_path": hand_row["original_path"],
        "handwritten_source_stem": hand_row["source_stem"],

        "ground_truth_row_idx": int(truth_row["row_idx"]),
        "ground_truth_raw_path": truth_row["raw_path"],
        "ground_truth_image_path": truth_row["rendered_image_path"],
        "ground_truth_original_path": truth_row["original_path"],
        "ground_truth_source_stem": truth_row["source_stem"],
        "ground_truth_text_path": truth_row["ground_truth_text_path"],
    })

# Create manifest DataFrame.
manifest_df = pd.DataFrame(manifest_rows)

if len(manifest_df) == 0:
    raise RuntimeError(
        "No pairs were created. Inspect processed_files.csv and check label names/filenames."
    )

# Fill record IDs.
manifest_df = manifest_df.sort_values(
    ["pair_confidence", "pair_method", "pairing_key", "handwritten_row_idx"],
    ascending=[True, True, True, True]
).reset_index(drop=True)

manifest_df["record_id"] = [f"AMR_{i+1:03d}" for i in range(len(manifest_df))]

# Load ground-truth text into manifest.
gt_texts = []
for p in manifest_df["ground_truth_text_path"]:
    p = Path(p)
    if p.exists():
        gt_texts.append(p.read_text(encoding="utf-8"))
    else:
        gt_texts.append("")

manifest_df["ground_truth_text"] = gt_texts
manifest_df["ground_truth_text_length"] = [len(x) for x in gt_texts]

manifest_path = PROJECT_DIR / "data" / "manifest.csv"
manifest_df.to_csv(manifest_path, index=False)

print("Manifest saved to:", manifest_path)
print("Number of paired records:", len(manifest_df))
print()
display(manifest_df.head(20))

print()
print("Pairing methods:")
display(manifest_df["pair_method"].value_counts().to_frame("count"))

print()
print("Pair confidence:")
display(manifest_df["pair_confidence"].value_counts().to_frame("count"))
````

````python
# ============================================================
# STEP 3.4 — Write amr_benchmark/text_utils.py
# ============================================================

%%writefile amr_benchmark/text_utils.py
from __future__ import annotations

import re
from typing import Optional


def safe_text(value: object) -> str:
    """
    Convert any value into a clean string.

    This avoids errors when a model returns None, NaN, or a non-string object.
    """
    if value is None:
        return ""

    text = str(value)

    # Pandas can stringify NaN as "nan"; treat that as missing only when exact.
    if text.strip().lower() == "nan":
        return ""

    return text


def normalize_line_endings(text: object) -> str:
    """
    Standardize line endings.
    """
    text = safe_text(text)
    return text.replace("\r\n", "\n").replace("\r", "\n")


def strip_code_fences(text: object) -> str:
    """
    Some models may return text inside Markdown code fences.
    Remove only the wrapper, not the actual content.
    """
    text = normalize_line_endings(text).strip()

    # Remove opening fence such as ``` or ```text
    text = re.sub(r"^```[a-zA-Z0-9_-]*\s*\n", "", text)

    # Remove closing fence
    text = re.sub(r"\n?```\s*$", "", text)

    return text.strip()


def remove_common_model_prefixes(text: object) -> str:
    """
    Remove common assistant-style preambles.

    The benchmark prompt asks for plain text only, but some models still say:
        'Here is the transcription:'

    We remove only obvious wrappers. We do not correct clinical content.
    """
    text = strip_code_fences(text)

    patterns = [
        r"^\s*here is the transcription\s*:\s*",
        r"^\s*transcription\s*:\s*",
        r"^\s*the transcription is\s*:\s*",
        r"^\s*plain text transcription\s*:\s*",
    ]

    for pattern in patterns:
        text = re.sub(pattern, "", text, flags=re.IGNORECASE)

    return text.strip()


def normalize_for_metric(text: object) -> str:
    """
    Normalize text for CER/WER calculation.

    Important:
    - Does NOT lowercase.
    - Does NOT remove punctuation.
    - Does NOT expand abbreviations.
    - Does NOT correct spelling.
    - Collapses whitespace so line-wrap differences do not dominate scores.
    """
    text = remove_common_model_prefixes(text)
    text = normalize_line_endings(text)

    # Replace any whitespace sequence with a single space.
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def preview_text(text: object, max_chars: int = 500) -> str:
    """
    Short preview for logs and notebook display.
    """
    text = safe_text(text)
    if len(text) <= max_chars:
        return text
    return text[:max_chars] + "..."
````

````python
# ============================================================
# STEP 3.5 — Write amr_benchmark/metrics.py
# ============================================================

%%writefile amr_benchmark/metrics.py
from __future__ import annotations

from typing import List, Sequence, Dict, Any
import pandas as pd

from .text_utils import normalize_for_metric, safe_text


def levenshtein_distance(a: Sequence, b: Sequence) -> int:
    """
    Compute Levenshtein edit distance between two sequences.

    This works for:
    - character sequences
    - word-token sequences

    Operations counted:
    - insertion
    - deletion
    - substitution
    """
    n = len(a)
    m = len(b)

    if n == 0:
        return m
    if m == 0:
        return n

    previous = list(range(m + 1))

    for i in range(1, n + 1):
        current = [i] + [0] * m

        for j in range(1, m + 1):
            deletion = previous[j] + 1
            insertion = current[j - 1] + 1
            substitution = previous[j - 1] + (0 if a[i - 1] == b[j - 1] else 1)

            current[j] = min(deletion, insertion, substitution)

        previous = current

    return previous[m]


def cer(reference: object, hypothesis: object) -> float:
    """
    Character Error Rate.

    CER = character-level edit distance / number of reference characters.

    Lower is better.
    0.0 means perfect character-level match after benchmark normalization.
    """
    ref = normalize_for_metric(reference)
    hyp = normalize_for_metric(hypothesis)

    if len(ref) == 0:
        return 0.0 if len(hyp) == 0 else 1.0

    return levenshtein_distance(ref, hyp) / len(ref)


def wer(reference: object, hypothesis: object) -> float:
    """
    Word Error Rate.

    WER = word-level edit distance / number of reference words.

    Lower is better.
    0.0 means perfect word-level match after benchmark normalization.
    """
    ref = normalize_for_metric(reference)
    hyp = normalize_for_metric(hypothesis)

    ref_words = ref.split()
    hyp_words = hyp.split()

    if len(ref_words) == 0:
        return 0.0 if len(hyp_words) == 0 else 1.0

    return levenshtein_distance(ref_words, hyp_words) / len(ref_words)


def compute_basic_metrics(reference: object, hypothesis: object) -> Dict[str, float]:
    """
    Compute the two main benchmark metrics.
    """
    return {
        "cer": cer(reference, hypothesis),
        "wer": wer(reference, hypothesis),
    }


def compute_span_cers(
    record_id: str,
    reference: object,
    hypothesis: object,
    field_spans_df: pd.DataFrame | None = None,
) -> Dict[str, float | None]:
    """
    Compute structured-content CER and narrative-content CER if span annotations exist.

    Expected field_spans.csv columns:
        record_id, span_type, start_char, end_char

    span_type should be either:
        structured
        narrative

    Notes:
    - Spans are interpreted as character offsets in the reference text.
    - If your team has not annotated these spans yet, this returns None values.
    - The hypothesis span is approximated by the same character range.
      This is a simple first-pass method and should be replaced by a stricter
      alignment method if your team needs high-precision span-level analysis.
    """
    if field_spans_df is None or len(field_spans_df) == 0:
        return {
            "struct_cer": None,
            "narr_cer": None,
        }

    required = {"record_id", "span_type", "start_char", "end_char"}
    if not required.issubset(set(field_spans_df.columns)):
        return {
            "struct_cer": None,
            "narr_cer": None,
        }

    ref = safe_text(reference)
    hyp = safe_text(hypothesis)

    subset = field_spans_df[field_spans_df["record_id"].astype(str) == str(record_id)].copy()

    if len(subset) == 0:
        return {
            "struct_cer": None,
            "narr_cer": None,
        }

    out = {}

    for span_label, out_col in [("structured", "struct_cer"), ("narrative", "narr_cer")]:
        span_rows = subset[subset["span_type"].astype(str).str.lower() == span_label]

        if len(span_rows) == 0:
            out[out_col] = None
            continue

        ref_parts = []
        hyp_parts = []

        for _, row in span_rows.iterrows():
            try:
                start = int(row["start_char"])
                end = int(row["end_char"])
            except Exception:
                continue

            if start < 0:
                start = 0
            if end < start:
                continue

            ref_parts.append(ref[start:end])
            hyp_parts.append(hyp[start:end])

        if len(ref_parts) == 0:
            out[out_col] = None
        else:
            out[out_col] = cer("\n".join(ref_parts), "\n".join(hyp_parts))

    return {
        "struct_cer": out.get("struct_cer"),
        "narr_cer": out.get("narr_cer"),
    }
````

````python
# ============================================================
# STEP 3.6 — Write amr_benchmark/io_utils.py
# ============================================================

%%writefile amr_benchmark/io_utils.py
from __future__ import annotations

import csv
import json
from pathlib import Path
from datetime import datetime, timezone
from typing import Dict, Any, Iterable, List, Tuple

import pandas as pd

from .config import ensure_project_dirs, project_paths
from .text_utils import safe_text


RESULT_COLUMNS = [
    "record_id",
    "model",
    "model_version",
    "run",
    "cer",
    "wer",
    "legibility_tier",
    "struct_cer",
    "narr_cer",
]

RAW_OUTPUT_COLUMNS = [
    "record_id",
    "model",
    "model_version",
    "run",
    "prompt",
    "raw_output",
    "clean_output",
    "ground_truth_text",
    "image_path",
    "created_at_utc",
]

FAILURE_COLUMNS = [
    "record_id",
    "model",
    "model_version",
    "run",
    "image_path",
    "error_type",
    "error_message",
    "created_at_utc",
]


def utc_now_iso() -> str:
    """
    Timestamp helper for reproducible logs.
    """
    return datetime.now(timezone.utc).isoformat()


def read_prompt(project_dir: str | Path | None = None) -> str:
    """
    Read the shared fixed prompt.
    """
    paths = project_paths(project_dir)
    prompt_path = paths["prompt_path"]

    if not prompt_path.exists():
        raise FileNotFoundError(
            f"Prompt file not found: {prompt_path}. "
            "Run Step 1 first."
        )

    return prompt_path.read_text(encoding="utf-8").strip()


def read_manifest(project_dir: str | Path | None = None) -> pd.DataFrame:
    """
    Load data/manifest.csv created in Step 2.
    """
    paths = project_paths(project_dir)
    manifest_path = paths["manifest_path"]

    if not manifest_path.exists():
        raise FileNotFoundError(
            f"Manifest file not found: {manifest_path}. "
            "Run Step 2 first."
        )

    df = pd.read_csv(manifest_path)

    required = {
        "record_id",
        "handwritten_image_path",
        "ground_truth_text",
    }

    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"Manifest is missing required columns: {sorted(missing)}")

    return df


def read_legibility(project_dir: str | Path | None = None) -> pd.DataFrame:
    """
    Load legibility metadata if available.

    Expected columns:
        record_id, legibility_tier

    If missing or empty, the caller can use 'unrated'.
    """
    paths = project_paths(project_dir)
    p = paths["legibility_path"]

    if not p.exists():
        return pd.DataFrame(columns=["record_id", "legibility_tier"])

    df = pd.read_csv(p)

    if "record_id" not in df.columns or "legibility_tier" not in df.columns:
        return pd.DataFrame(columns=["record_id", "legibility_tier"])

    return df


def read_field_spans(project_dir: str | Path | None = None) -> pd.DataFrame:
    """
    Load structured/narrative span metadata if available.
    """
    paths = project_paths(project_dir)
    p = paths["field_spans_path"]

    if not p.exists():
        return pd.DataFrame(columns=["record_id", "span_type", "start_char", "end_char"])

    return pd.read_csv(p)


def get_legibility_map(project_dir: str | Path | None = None) -> Dict[str, str]:
    """
    Return {record_id: legibility_tier}.
    """
    df = read_legibility(project_dir)

    if len(df) == 0:
        return {}

    out = {}

    for _, row in df.iterrows():
        rid = str(row.get("record_id", "")).strip()
        tier = str(row.get("legibility_tier", "")).strip()

        if rid:
            out[rid] = tier or "unrated"

    return out


def append_row_csv(path: str | Path, row: Dict[str, Any], columns: List[str]) -> None:
    """
    Append one row to a CSV, creating the file with header if needed.
    """
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)

    file_exists = path.exists()

    clean_row = {col: row.get(col, "") for col in columns}

    with path.open("a", encoding="utf-8", newline="") as f:
        writer = csv.DictWriter(f, fieldnames=columns)

        if not file_exists:
            writer.writeheader()

        writer.writerow(clean_row)


def append_raw_output(
    row: Dict[str, Any],
    project_dir: str | Path | None = None,
) -> None:
    """
    Append one raw model output row.
    """
    paths = project_paths(project_dir)
    append_row_csv(paths["raw_outputs_csv"], row, RAW_OUTPUT_COLUMNS)


def append_result(
    row: Dict[str, Any],
    project_dir: str | Path | None = None,
) -> None:
    """
    Append one scored result row.
    """
    paths = project_paths(project_dir)
    append_row_csv(paths["results_csv"], row, RESULT_COLUMNS)


def append_failure(
    row: Dict[str, Any],
    project_dir: str | Path | None = None,
) -> None:
    """
    Append one failure row.
    """
    paths = project_paths(project_dir)
    append_row_csv(paths["failures_csv"], row, FAILURE_COLUMNS)


def save_text_output(
    record_id: str,
    model: str,
    run: int,
    text: object,
    project_dir: str | Path | None = None,
) -> Path:
    """
    Save a single model transcription as a .txt file.

    This makes debugging easier than relying only on CSV files.
    """
    paths = project_paths(project_dir)

    model_safe = str(model).replace("/", "_").replace(" ", "_")
    out_dir = paths["outputs_dir"] / model_safe
    out_dir.mkdir(parents=True, exist_ok=True)

    out_path = out_dir / f"{record_id}_run{int(run)}.txt"
    out_path.write_text(safe_text(text), encoding="utf-8")

    return out_path


def load_existing_result_keys(project_dir: str | Path | None = None) -> set[Tuple[str, str, int]]:
    """
    Return already-completed result keys:
        (record_id, model, run)

    Used for --resume logic later.
    """
    paths = project_paths(project_dir)
    p = paths["results_csv"]

    if not p.exists():
        return set()

    df = pd.read_csv(p)

    if len(df) == 0:
        return set()

    keys = set()

    for _, row in df.iterrows():
        try:
            keys.add((str(row["record_id"]), str(row["model"]), int(row["run"])))
        except Exception:
            continue

    return keys


def initialize_output_files(project_dir: str | Path | None = None) -> None:
    """
    Ensure output folders exist.

    This does not erase existing results.
    """
    ensure_project_dirs(project_dir)
````

````python
# ============================================================
# STEP 3.7 — Write amr_benchmark/evaluation.py
# ============================================================

%%writefile amr_benchmark/evaluation.py
from __future__ import annotations

from pathlib import Path
from typing import Dict, Any

import pandas as pd

from .metrics import compute_basic_metrics, compute_span_cers
from .text_utils import remove_common_model_prefixes
from .io_utils import (
    get_legibility_map,
    read_field_spans,
)


def score_single_output(
    record_id: str,
    model: str,
    model_version: str,
    run: int,
    ground_truth_text: str,
    raw_output: str,
    project_dir: str | Path | None = None,
) -> Dict[str, Any]:
    """
    Score one model output against one ground-truth transcription.

    Returns a row compatible with:
        results/raw/results_opensource_member2.csv
    """
    clean_output = remove_common_model_prefixes(raw_output)

    basic = compute_basic_metrics(
        reference=ground_truth_text,
        hypothesis=clean_output,
    )

    field_spans_df = read_field_spans(project_dir)
    span_metrics = compute_span_cers(
        record_id=record_id,
        reference=ground_truth_text,
        hypothesis=clean_output,
        field_spans_df=field_spans_df,
    )

    legibility_map = get_legibility_map(project_dir)
    legibility_tier = legibility_map.get(str(record_id), "unrated")

    return {
        "record_id": record_id,
        "model": model,
        "model_version": model_version,
        "run": int(run),
        "cer": basic["cer"],
        "wer": basic["wer"],
        "legibility_tier": legibility_tier,
        "struct_cer": span_metrics["struct_cer"],
        "narr_cer": span_metrics["narr_cer"],
    }


def rebuild_results_from_raw_outputs(project_dir: str | Path) -> pd.DataFrame:
    """
    Recompute results_opensource_member2.csv from raw_model_outputs_member2.csv.

    Use this if:
    - you changed metric normalization;
    - you corrected ground-truth text;
    - you want to regenerate the final scored CSV without rerunning models.
    """
    from .config import project_paths

    paths = project_paths(project_dir)
    raw_path = paths["raw_outputs_csv"]
    results_path = paths["results_csv"]

    if not raw_path.exists():
        raise FileNotFoundError(f"Raw outputs CSV not found: {raw_path}")

    raw_df = pd.read_csv(raw_path)

    rows = []

    for _, row in raw_df.iterrows():
        scored = score_single_output(
            record_id=str(row["record_id"]),
            model=str(row["model"]),
            model_version=str(row["model_version"]),
            run=int(row["run"]),
            ground_truth_text=str(row["ground_truth_text"]),
            raw_output=str(row["raw_output"]),
            project_dir=project_dir,
        )
        rows.append(scored)

    result_df = pd.DataFrame(rows)
    result_df.to_csv(results_path, index=False)

    return result_df
````

````python
%%writefile amr_benchmark/model_adapters.py
from __future__ import annotations

import os
import gc
import json
import base64
import tempfile
from pathlib import Path
from typing import Any, Dict, Optional

import torch
from PIL import Image


# ============================================================
# Shared helpers
# ============================================================

def get_best_torch_dtype() -> torch.dtype:
    """
    Pick a reasonable GPU dtype.

    bfloat16 is preferred on newer NVIDIA GPUs.
    float16 is used as fallback.
    """
    if not torch.cuda.is_available():
        return torch.float32

    major, _minor = torch.cuda.get_device_capability(0)

    if major >= 8:
        return torch.bfloat16

    return torch.float16


def clear_gpu_memory() -> None:
    """
    Free GPU memory between model runs.
    """
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.ipc_collect()


def ensure_image_path(image_path: str | Path) -> Path:
    """
    Validate image path before model inference.
    """
    image_path = Path(image_path)

    if not image_path.exists():
        raise FileNotFoundError(f"Image not found: {image_path}")

    return image_path


def load_rgb_image(image_path: str | Path) -> Image.Image:
    """
    Load image as RGB.
    """
    image_path = ensure_image_path(image_path)
    return Image.open(image_path).convert("RGB")


def encode_image_to_data_url(image_path: str | Path) -> str:
    """
    Encode image as a base64 data URL for APIs that accept image input.
    """
    image_path = ensure_image_path(image_path)
    suffix = image_path.suffix.lower()

    if suffix in [".jpg", ".jpeg"]:
        mime = "image/jpeg"
    else:
        mime = "image/png"

    data = image_path.read_bytes()
    b64 = base64.b64encode(data).decode("utf-8")

    return f"data:{mime};base64,{b64}"


def safe_model_output(text: Any) -> str:
    """
    Convert model output to string safely.
    """
    if text is None:
        return ""

    return str(text).strip()


# ============================================================
# Base class
# ============================================================

class BaseModelAdapter:
    """
    Base class for all benchmark model adapters.

    Every adapter must expose:
        - model_name
        - model_version
        - load()
        - generate(image_path, prompt, max_new_tokens)
        - unload()
    """

    model_name: str = "base"
    model_version: str = "unknown"

    def __init__(
        self,
        model_version: Optional[str] = None,
        device_map: str | Dict[str, Any] = "auto",
        torch_dtype: Optional[torch.dtype] = None,
        use_4bit: bool = False,
    ) -> None:
        self.model_version = model_version or self.model_version
        self.device_map = device_map
        self.torch_dtype = torch_dtype or get_best_torch_dtype()
        self.use_4bit = use_4bit

        self.model = None
        self.processor = None
        self.tokenizer = None
        self.loaded = False

    def load(self) -> "BaseModelAdapter":
        raise NotImplementedError

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        raise NotImplementedError

    def unload(self) -> None:
        """
        Delete large objects and free GPU memory.
        """
        self.model = None
        self.processor = None
        self.tokenizer = None
        self.loaded = False
        clear_gpu_memory()


# ============================================================
# GPT-4o adapter
# ============================================================

class GPT4oAdapter(BaseModelAdapter):
    """
    Adapter for GPT-4o through the OpenAI API.

    Required environment variable:
        OPENAI_API_KEY

    This adapter sends the handwritten record image and the fixed prompt.
    """

    model_name = "gpt4o"
    model_version = "gpt-4o"

    def __init__(self, model_version: str = "gpt-4o", **kwargs: Any) -> None:
        super().__init__(model_version=model_version, **kwargs)
        self.client = None

    def load(self) -> "GPT4oAdapter":
        if not os.environ.get("OPENAI_API_KEY"):
            raise EnvironmentError(
                "OPENAI_API_KEY is not set. "
                "Set it before running GPT-4o, for example:\n"
                'os.environ["OPENAI_API_KEY"] = "sk-..."'
            )

        from openai import OpenAI

        self.client = OpenAI()
        self.loaded = True
        return self

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        if not self.loaded:
            self.load()

        image_url = encode_image_to_data_url(image_path)

        response = self.client.responses.create(
            model=self.model_version,
            input=[
                {
                    "role": "user",
                    "content": [
                        {
                            "type": "input_text",
                            "text": prompt,
                        },
                        {
                            "type": "input_image",
                            "image_url": image_url,
                        },
                    ],
                }
            ],
            temperature=0,
            max_output_tokens=max_new_tokens,
        )

        return safe_model_output(response.output_text)

    def unload(self) -> None:
        self.client = None
        self.loaded = False


# ============================================================
# Qwen3-VL adapter
# ============================================================

class Qwen3VLAdapter(BaseModelAdapter):
    """
    Adapter for Qwen3-VL.

    Default checkpoint:
        Qwen/Qwen3-VL-8B-Instruct

    For smaller GPU memory, you may later switch to:
        Qwen/Qwen3-VL-4B-Instruct
        Qwen/Qwen3-VL-2B-Instruct

    The benchmark should record the exact model_version used.
    """

    model_name = "qwen3_vl"
    model_version = "Qwen/Qwen3-VL-8B-Instruct"

    def load(self) -> "Qwen3VLAdapter":
        from transformers import AutoProcessor

        try:
            from transformers import Qwen3VLForConditionalGeneration
            model_cls = Qwen3VLForConditionalGeneration
        except Exception:
            # Fallback for some Transformers versions.
            from transformers import AutoModelForImageTextToText
            model_cls = AutoModelForImageTextToText

        quantization_config = None

        if self.use_4bit:
            from transformers import BitsAndBytesConfig

            quantization_config = BitsAndBytesConfig(
                load_in_4bit=True,
                bnb_4bit_compute_dtype=self.torch_dtype,
                bnb_4bit_quant_type="nf4",
                bnb_4bit_use_double_quant=True,
            )

        self.model = model_cls.from_pretrained(
            self.model_version,
            torch_dtype=self.torch_dtype,
            device_map=self.device_map,
            trust_remote_code=True,
            quantization_config=quantization_config,
        )

        self.processor = AutoProcessor.from_pretrained(
            self.model_version,
            trust_remote_code=True,
        )

        self.model.eval()
        self.loaded = True
        return self

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        if not self.loaded:
            self.load()

        from qwen_vl_utils import process_vision_info

        image_path = ensure_image_path(image_path)

        messages = [
            {
                "role": "user",
                "content": [
                    {
                        "type": "image",
                        "image": str(image_path),
                    },
                    {
                        "type": "text",
                        "text": prompt,
                    },
                ],
            }
        ]

        text = self.processor.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True,
        )

        image_inputs, video_inputs = process_vision_info(messages)

        inputs = self.processor(
            text=[text],
            images=image_inputs,
            videos=video_inputs,
            padding=True,
            return_tensors="pt",
        )

        # Move tensors to model device.
        inputs = {
            k: v.to(self.model.device) if torch.is_tensor(v) else v
            for k, v in inputs.items()
        }

        with torch.inference_mode():
            generated_ids = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
            )

        # Remove prompt tokens from generated output.
        input_len = inputs["input_ids"].shape[1]
        generated_trimmed = generated_ids[:, input_len:]

        output_text = self.processor.batch_decode(
            generated_trimmed,
            skip_special_tokens=True,
            clean_up_tokenization_spaces=False,
        )[0]

        return safe_model_output(output_text)


# ============================================================
# dots.ocr adapter
# ============================================================

class DotsOCRAdapter(BaseModelAdapter):
    """
    Adapter for rednote-hilab/dots.ocr.

    dots.ocr is a document parser/OCR VLM. This adapter uses a generic
    Transformers + trust_remote_code loading pattern. During smoke testing,
    this is the first adapter to verify because some dots.ocr releases may
    require repository-specific helper code.
    """

    model_name = "dots_ocr"
    model_version = "rednote-hilab/dots.ocr"

    def load(self) -> "DotsOCRAdapter":
        from transformers import AutoProcessor, AutoModelForCausalLM

        quantization_config = None

        if self.use_4bit:
            from transformers import BitsAndBytesConfig

            quantization_config = BitsAndBytesConfig(
                load_in_4bit=True,
                bnb_4bit_compute_dtype=self.torch_dtype,
                bnb_4bit_quant_type="nf4",
                bnb_4bit_use_double_quant=True,
            )

        self.model = AutoModelForCausalLM.from_pretrained(
            self.model_version,
            torch_dtype=self.torch_dtype,
            device_map=self.device_map,
            trust_remote_code=True,
            quantization_config=quantization_config,
        )

        self.processor = AutoProcessor.from_pretrained(
            self.model_version,
            trust_remote_code=True,
        )

        self.model.eval()
        self.loaded = True
        return self

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        if not self.loaded:
            self.load()

        image_path = ensure_image_path(image_path)

        # First try chat-style API if the remote model exposes one.
        if hasattr(self.model, "chat"):
            image = load_rgb_image(image_path)

            try:
                response = self.model.chat(
                    image=image,
                    msgs=[{"role": "user", "content": prompt}],
                    tokenizer=self.processor,
                    sampling=False,
                    temperature=0,
                    max_new_tokens=max_new_tokens,
                )
                return safe_model_output(response)
            except TypeError:
                # Fall through to generic generation.
                pass

        # Generic VLM generation path.
        image = load_rgb_image(image_path)

        messages = [
            {
                "role": "user",
                "content": [
                    {
                        "type": "image",
                    },
                    {
                        "type": "text",
                        "text": prompt,
                    },
                ],
            }
        ]

        try:
            chat_text = self.processor.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True,
            )
        except Exception:
            chat_text = prompt

        inputs = self.processor(
            text=chat_text,
            images=image,
            return_tensors="pt",
        )

        inputs = {
            k: v.to(self.model.device) if torch.is_tensor(v) else v
            for k, v in inputs.items()
        }

        with torch.inference_mode():
            generated_ids = self.model.generate(
                **inputs,
                max_new_tokens=max_new_tokens,
                do_sample=False,
            )

        input_len = inputs["input_ids"].shape[1]
        generated_trimmed = generated_ids[:, input_len:]

        try:
            output_text = self.processor.batch_decode(
                generated_trimmed,
                skip_special_tokens=True,
                clean_up_tokenization_spaces=False,
            )[0]
        except Exception:
            output_text = self.processor.decode(
                generated_trimmed[0],
                skip_special_tokens=True,
            )

        return safe_model_output(output_text)


# ============================================================
# InternVL3 adapter
# ============================================================

class InternVL3Adapter(BaseModelAdapter):
    """
    Adapter for InternVL3.

    Default checkpoint:
        OpenGVLab/InternVL3-8B

    InternVL uses a repository-specific chat method and image preprocessing.
    """

    model_name = "internvl3"
    model_version = "OpenGVLab/InternVL3-8B"

    def load(self) -> "InternVL3Adapter":
        from transformers import AutoTokenizer, AutoModel

        quantization_config = None

        if self.use_4bit:
            from transformers import BitsAndBytesConfig

            quantization_config = BitsAndBytesConfig(
                load_in_4bit=True,
                bnb_4bit_compute_dtype=self.torch_dtype,
                bnb_4bit_quant_type="nf4",
                bnb_4bit_use_double_quant=True,
            )

        self.tokenizer = AutoTokenizer.from_pretrained(
            self.model_version,
            trust_remote_code=True,
            use_fast=False,
        )

        self.model = AutoModel.from_pretrained(
            self.model_version,
            torch_dtype=self.torch_dtype,
            low_cpu_mem_usage=True,
            trust_remote_code=True,
            device_map=self.device_map,
            quantization_config=quantization_config,
        )

        self.model.eval()
        self.loaded = True
        return self

    @staticmethod
    def _build_transform(input_size: int = 448):
        import torchvision.transforms as T
        from torchvision.transforms.functional import InterpolationMode

        imagenet_mean = (0.485, 0.456, 0.406)
        imagenet_std = (0.229, 0.224, 0.225)

        transform = T.Compose(
            [
                T.Lambda(lambda img: img.convert("RGB") if img.mode != "RGB" else img),
                T.Resize((input_size, input_size), interpolation=InterpolationMode.BICUBIC),
                T.ToTensor(),
                T.Normalize(mean=imagenet_mean, std=imagenet_std),
            ]
        )

        return transform

    @staticmethod
    def _find_closest_aspect_ratio(
        aspect_ratio: float,
        target_ratios: list[tuple[int, int]],
        width: int,
        height: int,
        image_size: int,
    ) -> tuple[int, int]:
        best_ratio_diff = float("inf")
        best_ratio = (1, 1)
        area = width * height

        for ratio in target_ratios:
            target_aspect_ratio = ratio[0] / ratio[1]
            ratio_diff = abs(aspect_ratio - target_aspect_ratio)

            if ratio_diff < best_ratio_diff:
                best_ratio_diff = ratio_diff
                best_ratio = ratio
            elif ratio_diff == best_ratio_diff:
                if area > 0.5 * image_size * image_size * ratio[0] * ratio[1]:
                    best_ratio = ratio

        return best_ratio

    @classmethod
    def _dynamic_preprocess(
        cls,
        image: Image.Image,
        min_num: int = 1,
        max_num: int = 12,
        image_size: int = 448,
        use_thumbnail: bool = True,
    ) -> list[Image.Image]:
        """
        Split a large page into tiles for InternVL.
        """
        orig_width, orig_height = image.size
        aspect_ratio = orig_width / orig_height

        target_ratios = set(
            (i, j)
            for n in range(min_num, max_num + 1)
            for i in range(1, n + 1)
            for j in range(1, n + 1)
            if i * j <= max_num and i * j >= min_num
        )

        target_ratios = sorted(target_ratios, key=lambda x: x[0] * x[1])

        target_aspect_ratio = cls._find_closest_aspect_ratio(
            aspect_ratio,
            target_ratios,
            orig_width,
            orig_height,
            image_size,
        )

        target_width = image_size * target_aspect_ratio[0]
        target_height = image_size * target_aspect_ratio[1]
        blocks = target_aspect_ratio[0] * target_aspect_ratio[1]

        resized_img = image.resize((target_width, target_height))
        processed_images = []

        for i in range(blocks):
            box = (
                (i % (target_width // image_size)) * image_size,
                (i // (target_width // image_size)) * image_size,
                ((i % (target_width // image_size)) + 1) * image_size,
                ((i // (target_width // image_size)) + 1) * image_size,
            )
            split_img = resized_img.crop(box)
            processed_images.append(split_img)

        if use_thumbnail and len(processed_images) != 1:
            thumbnail_img = image.resize((image_size, image_size))
            processed_images.append(thumbnail_img)

        return processed_images

    @classmethod
    def _load_image_tensor(
        cls,
        image_path: str | Path,
        input_size: int = 448,
        max_num: int = 12,
    ) -> torch.Tensor:
        image = load_rgb_image(image_path)
        transform = cls._build_transform(input_size=input_size)

        images = cls._dynamic_preprocess(
            image,
            image_size=input_size,
            use_thumbnail=True,
            max_num=max_num,
        )

        pixel_values = [transform(img) for img in images]
        pixel_values = torch.stack(pixel_values)

        return pixel_values

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        if not self.loaded:
            self.load()

        pixel_values = self._load_image_tensor(
            image_path,
            input_size=448,
            max_num=12,
        )

        pixel_values = pixel_values.to(dtype=self.torch_dtype)

        if torch.cuda.is_available():
            pixel_values = pixel_values.cuda()

        question = "<image>\n" + prompt

        generation_config = {
            "max_new_tokens": max_new_tokens,
            "do_sample": False,
        }

        with torch.inference_mode():
            response = self.model.chat(
                self.tokenizer,
                pixel_values,
                question,
                generation_config,
            )

        return safe_model_output(response)


# ============================================================
# PaddleOCR-VL adapter
# ============================================================

class PaddleOCRVLAdapter(BaseModelAdapter):
    """
    Adapter for PaddleOCR-VL through the PaddleOCR package.

    PaddleOCR-VL is a document parsing model. It may return structured
    objects rather than plain text, so this adapter recursively extracts
    likely text fields.
    """

    model_name = "paddleocr_vl"
    model_version = "PaddlePaddle/PaddleOCR-VL"

    def __init__(self, model_version: str = "PaddlePaddle/PaddleOCR-VL", **kwargs: Any) -> None:
        super().__init__(model_version=model_version, **kwargs)
        self.pipeline = None

    def load(self) -> "PaddleOCRVLAdapter":
        try:
            from paddleocr import PaddleOCRVL
        except Exception as e:
            raise ImportError(
                "Could not import PaddleOCRVL from paddleocr. "
                "Try restarting the runtime after installing paddleocr, then run Step 4 again. "
                f"Original error: {repr(e)}"
            )

        self.pipeline = PaddleOCRVL()
        self.loaded = True
        return self

    @staticmethod
    def _extract_text_recursive(obj: Any) -> list[str]:
        """
        Recursively extract text-like fields from PaddleOCR output.
        """
        texts = []

        if obj is None:
            return texts

        if isinstance(obj, str):
            s = obj.strip()
            if s:
                texts.append(s)
            return texts

        if isinstance(obj, (int, float, bool)):
            return texts

        if isinstance(obj, dict):
            preferred_keys = [
                "text",
                "rec_text",
                "rec_texts",
                "transcription",
                "content",
                "markdown",
                "html",
                "plain_text",
                "res",
                "parsing_res_list",
            ]

            for key in preferred_keys:
                if key in obj:
                    texts.extend(PaddleOCRVLAdapter._extract_text_recursive(obj[key]))

            # Also scan all remaining fields, but avoid duplicates later.
            for key, value in obj.items():
                if key not in preferred_keys:
                    texts.extend(PaddleOCRVLAdapter._extract_text_recursive(value))

            return texts

        if isinstance(obj, (list, tuple)):
            for item in obj:
                texts.extend(PaddleOCRVLAdapter._extract_text_recursive(item))
            return texts

        # Some Paddle result objects expose JSON-like attributes/methods.
        for attr in ["json", "to_dict", "dict"]:
            if hasattr(obj, attr):
                try:
                    value = getattr(obj, attr)
                    value = value() if callable(value) else value
                    texts.extend(PaddleOCRVLAdapter._extract_text_recursive(value))
                    return texts
                except Exception:
                    pass

        # Fallback string representation.
        s = str(obj).strip()
        if s and not s.startswith("<") and not s.endswith(">"):
            texts.append(s)

        return texts

    @staticmethod
    def _dedupe_lines(lines: list[str]) -> str:
        seen = set()
        out = []

        for line in lines:
            line = str(line).strip()
            if not line:
                continue

            if line in seen:
                continue

            seen.add(line)
            out.append(line)

        return "\n".join(out).strip()

    def generate(
        self,
        image_path: str | Path,
        prompt: str,
        max_new_tokens: int = 2048,
    ) -> str:
        if not self.loaded:
            self.load()

        image_path = ensure_image_path(image_path)

        # PaddleOCR-VL does not necessarily use the benchmark prompt directly.
        # It is an OCR/document parser. We still pass the same image and extract
        # plain text from its output.
        result = self.pipeline.predict(input=str(image_path))

        text_parts = self._extract_text_recursive(result)
        text = self._dedupe_lines(text_parts)

        return safe_model_output(text)

    def unload(self) -> None:
        self.pipeline = None
        self.loaded = False
        clear_gpu_memory()


# ============================================================
# Adapter registry
# ============================================================

MODEL_REGISTRY = {
    "gpt4o": GPT4oAdapter,
    "qwen3_vl": Qwen3VLAdapter,
    "dots_ocr": DotsOCRAdapter,
    "internvl3": InternVL3Adapter,
    "paddleocr_vl": PaddleOCRVLAdapter,
}


def available_models() -> list[str]:
    """
    Return model names available in this codebase.
    """
    return sorted(MODEL_REGISTRY.keys())


def get_model_adapter(
    model_name: str,
    model_version: Optional[str] = None,
    use_4bit: bool = False,
    device_map: str | Dict[str, Any] = "auto",
) -> BaseModelAdapter:
    """
    Create a model adapter by name.

    This does not load the model immediately.
    Loading happens when adapter.load() or adapter.generate() is called.
    """
    key = str(model_name).strip().lower()

    if key not in MODEL_REGISTRY:
        raise KeyError(
            f"Unknown model '{model_name}'. "
            f"Available models: {available_models()}"
        )

    cls = MODEL_REGISTRY[key]

    kwargs = {
        "use_4bit": use_4bit,
        "device_map": device_map,
    }

    if model_version is not None:
        kwargs["model_version"] = model_version

    return cls(**kwargs)
````

````python
# ============================================================
# STEP 5.4 — Create smoke-test runner
# ============================================================

%%writefile run_smoke.py
from __future__ import annotations

import argparse
import traceback
from pathlib import Path
from datetime import datetime, timezone

import pandas as pd

from amr_benchmark.config import ensure_project_dirs, project_paths
from amr_benchmark.io_utils import read_manifest, read_prompt
from amr_benchmark.model_adapters import get_model_adapter, clear_gpu_memory
from amr_benchmark.text_utils import remove_common_model_prefixes, preview_text
from amr_benchmark.evaluation import score_single_output


def utc_now_iso() -> str:
    return datetime.now(timezone.utc).isoformat()


def append_csv(path: Path, row: dict) -> None:
    """
    Append one row to a CSV file, creating it if needed.
    """
    path.parent.mkdir(parents=True, exist_ok=True)

    df = pd.DataFrame([row])

    if path.exists():
        df.to_csv(path, mode="a", header=False, index=False)
    else:
        df.to_csv(path, index=False)


def select_record(manifest_df: pd.DataFrame, record_id: str | None, record_index: int) -> pd.Series:
    """
    Select one record either by record_id or by row index.
    """
    if record_id:
        subset = manifest_df[manifest_df["record_id"].astype(str) == str(record_id)]

        if len(subset) == 0:
            raise ValueError(f"record_id not found in manifest: {record_id}")

        return subset.iloc[0]

    if record_index < 0 or record_index >= len(manifest_df):
        raise IndexError(
            f"record_index {record_index} is out of range. "
            f"Manifest has {len(manifest_df)} records."
        )

    return manifest_df.iloc[record_index]


def main() -> None:
    parser = argparse.ArgumentParser(description="Smoke test one AMR benchmark model on one record.")

    parser.add_argument(
        "--model",
        required=True,
        help="Model adapter name: qwen3_vl, internvl3, dots_ocr, paddleocr_vl, or gpt4o.",
    )

    parser.add_argument(
        "--model-version",
        default=None,
        help="Optional exact checkpoint/API model override.",
    )

    parser.add_argument(
        "--record-id",
        default=None,
        help="Optional record_id from manifest, e.g. AMR_001.",
    )

    parser.add_argument(
        "--record-index",
        type=int,
        default=0,
        help="Manifest row index to test if --record-id is not supplied.",
    )

    parser.add_argument(
        "--run",
        type=int,
        default=1,
        help="Run number for smoke-test scoring display.",
    )

    parser.add_argument(
        "--max-new-tokens",
        type=int,
        default=1024,
        help="Maximum generation length for the smoke test.",
    )

    parser.add_argument(
        "--use-4bit",
        action="store_true",
        help="Load supported Hugging Face models in 4-bit quantization.",
    )

    parser.add_argument(
        "--write-final-results",
        action="store_true",
        help=(
            "If set, append this smoke-test result to the final results CSV. "
            "Normally leave this off so smoke tests do not pollute final results."
        ),
    )

    parser.add_argument(
        "--project-dir",
        default="/content/drive/MyDrive/amr_member2_benchmark",
        help="Main project directory.",
    )

    args = parser.parse_args()

    project_dir = Path(args.project_dir)
    ensure_project_dirs(project_dir)
    paths = project_paths(project_dir)

    smoke_log_path = project_dir / "logs" / "smoke_tests.csv"
    smoke_failure_path = project_dir / "logs" / "smoke_failures.csv"
    smoke_output_dir = project_dir / "logs" / "smoke_outputs"
    smoke_output_dir.mkdir(parents=True, exist_ok=True)

    manifest_df = read_manifest(project_dir)
    prompt = read_prompt(project_dir)

    row = select_record(
        manifest_df=manifest_df,
        record_id=args.record_id,
        record_index=args.record_index,
    )

    record_id = str(row["record_id"])
    image_path = Path(str(row["handwritten_image_path"]))
    ground_truth_text = str(row["ground_truth_text"])

    if not image_path.exists():
        raise FileNotFoundError(f"Handwritten image not found: {image_path}")

    adapter = None

    print("=" * 80)
    print("AMR SMOKE TEST")
    print("=" * 80)
    print("Model:", args.model)
    print("Model version override:", args.model_version)
    print("Use 4-bit:", args.use_4bit)
    print("Record:", record_id)
    print("Image:", image_path)
    print("Ground-truth characters:", len(ground_truth_text))
    print("Max new tokens:", args.max_new_tokens)
    print("=" * 80)

    try:
        clear_gpu_memory()

        adapter = get_model_adapter(
            model_name=args.model,
            model_version=args.model_version,
            use_4bit=args.use_4bit,
        )

        print("\nAdapter created:")
        print("model_name:", adapter.model_name)
        print("model_version:", adapter.model_version)

        print("\nLoading model...")
        adapter.load()
        print("Model loaded.")

        print("\nGenerating transcription...")
        raw_output = adapter.generate(
            image_path=image_path,
            prompt=prompt,
            max_new_tokens=args.max_new_tokens,
        )

        clean_output = remove_common_model_prefixes(raw_output)

        print("\nGeneration complete.")
        print("Raw output characters:", len(str(raw_output)))
        print("Clean output characters:", len(str(clean_output)))

        scored = score_single_output(
            record_id=record_id,
            model=adapter.model_name,
            model_version=adapter.model_version,
            run=args.run,
            ground_truth_text=ground_truth_text,
            raw_output=raw_output,
            project_dir=project_dir,
        )

        output_path = smoke_output_dir / f"{adapter.model_name}_{record_id}_run{args.run}.txt"
        output_path.write_text(str(raw_output), encoding="utf-8")

        smoke_row = {
            "created_at_utc": utc_now_iso(),
            "status": "success",
            "record_id": record_id,
            "model": adapter.model_name,
            "model_version": adapter.model_version,
            "run": args.run,
            "cer": scored["cer"],
            "wer": scored["wer"],
            "image_path": str(image_path),
            "output_path": str(output_path),
            "use_4bit": args.use_4bit,
            "max_new_tokens": args.max_new_tokens,
        }

        append_csv(smoke_log_path, smoke_row)

        if args.write_final_results:
            from amr_benchmark.io_utils import append_result, append_raw_output, save_text_output
            from amr_benchmark.io_utils import utc_now_iso as io_utc_now_iso

            append_result(scored, project_dir=project_dir)

            append_raw_output(
                {
                    "record_id": record_id,
                    "model": adapter.model_name,
                    "model_version": adapter.model_version,
                    "run": args.run,
                    "prompt": prompt,
                    "raw_output": raw_output,
                    "clean_output": clean_output,
                    "ground_truth_text": ground_truth_text,
                    "image_path": str(image_path),
                    "created_at_utc": io_utc_now_iso(),
                },
                project_dir=project_dir,
            )

            save_text_output(
                record_id=record_id,
                model=adapter.model_name,
                run=args.run,
                text=raw_output,
                project_dir=project_dir,
            )

        print("\n" + "=" * 80)
        print("SMOKE TEST RESULT")
        print("=" * 80)
        print("CER:", scored["cer"])
        print("WER:", scored["wer"])
        print("Legibility tier:", scored["legibility_tier"])
        print("Output saved to:", output_path)
        print("Smoke log saved to:", smoke_log_path)

        print("\n" + "=" * 80)
        print("MODEL OUTPUT PREVIEW")
        print("=" * 80)
        print(preview_text(raw_output, max_chars=2000))

        print("\n" + "=" * 80)
        print("GROUND TRUTH PREVIEW")
        print("=" * 80)
        print(preview_text(ground_truth_text, max_chars=2000))

    except Exception as e:
        err = traceback.format_exc()

        failure_row = {
            "created_at_utc": utc_now_iso(),
            "status": "failed",
            "record_id": record_id,
            "model": args.model,
            "model_version": args.model_version or "",
            "run": args.run,
            "image_path": str(image_path),
            "use_4bit": args.use_4bit,
            "max_new_tokens": args.max_new_tokens,
            "error_type": type(e).__name__,
            "error_message": str(e),
            "traceback": err,
        }

        append_csv(smoke_failure_path, failure_row)

        print("\n" + "=" * 80)
        print("SMOKE TEST FAILED")
        print("=" * 80)
        print(err)
        print("Failure log saved to:", smoke_failure_path)

        raise

    finally:
        if adapter is not None:
            adapter.unload()

        clear_gpu_memory()


if __name__ == "__main__":
    main()
````

````python
# ============================================================
# STEP 6.2 — Create full inference runner
# ============================================================

%%writefile run_inference.py
from __future__ import annotations

import argparse
import traceback
from pathlib import Path
from datetime import datetime, timezone
from typing import Optional

import pandas as pd

from amr_benchmark.config import ensure_project_dirs, project_paths
from amr_benchmark.io_utils import (
    read_manifest,
    read_prompt,
    append_result,
    append_raw_output,
    append_failure,
    save_text_output,
    load_existing_result_keys,
    utc_now_iso,
)
from amr_benchmark.model_adapters import (
    get_model_adapter,
    available_models,
    clear_gpu_memory,
)
from amr_benchmark.text_utils import remove_common_model_prefixes
from amr_benchmark.evaluation import score_single_output


def parse_models(models_arg: list[str]) -> list[str]:
    """
    Parse model list.

    Allows:
        --models qwen3_vl
        --models qwen3_vl internvl3
        --models all
    """
    if len(models_arg) == 1 and models_arg[0].lower() == "all":
        return available_models()

    return [m.strip().lower() for m in models_arg if m.strip()]


def subset_manifest(
    manifest_df: pd.DataFrame,
    record_id: Optional[str] = None,
    start_index: Optional[int] = None,
    end_index: Optional[int] = None,
    limit: Optional[int] = None,
) -> pd.DataFrame:
    """
    Select records for this run.

    Useful for:
    - testing first few records;
    - splitting long runs into batches;
    - rerunning one failed record.
    """
    df = manifest_df.copy()

    if record_id:
        df = df[df["record_id"].astype(str) == str(record_id)].copy()
        if len(df) == 0:
            raise ValueError(f"record_id not found in manifest: {record_id}")

    else:
        if start_index is not None or end_index is not None:
            start = 0 if start_index is None else int(start_index)
            end = len(df) if end_index is None else int(end_index)
            df = df.iloc[start:end].copy()

        if limit is not None:
            df = df.head(int(limit)).copy()

    return df.reset_index(drop=True)


def append_progress_log(project_dir: Path, row: dict) -> None:
    """
    Lightweight progress log for long Colab runs.
    """
    log_path = project_dir / "logs" / "inference_progress.csv"
    log_path.parent.mkdir(parents=True, exist_ok=True)

    df = pd.DataFrame([row])

    if log_path.exists():
        df.to_csv(log_path, mode="a", header=False, index=False)
    else:
        df.to_csv(log_path, index=False)


def main() -> None:
    parser = argparse.ArgumentParser(
        description="Run AMR Member 2 full model inference and scoring."
    )

    parser.add_argument(
        "--models",
        nargs="+",
        required=True,
        help=(
            "Model names to run. Options: "
            "qwen3_vl, internvl3, dots_ocr, paddleocr_vl, gpt4o, or all."
        ),
    )

    parser.add_argument(
        "--runs",
        type=int,
        default=3,
        help="Number of independent runs per record per model. Project default is 3.",
    )

    parser.add_argument(
        "--max-new-tokens",
        type=int,
        default=2048,
        help="Maximum output tokens per record.",
    )

    parser.add_argument(
        "--project-dir",
        default="/content/drive/MyDrive/amr_member2_benchmark",
        help="Main project directory.",
    )

    parser.add_argument(
        "--resume",
        action="store_true",
        help="Skip record/model/run combinations already present in final results CSV.",
    )

    parser.add_argument(
        "--use-4bit",
        action="store_true",
        help="Load supported Hugging Face models in 4-bit quantization.",
    )

    parser.add_argument(
        "--record-id",
        default=None,
        help="Run one specific record_id, e.g. AMR_001.",
    )

    parser.add_argument(
        "--start-index",
        type=int,
        default=None,
        help="Start row index in manifest, inclusive.",
    )

    parser.add_argument(
        "--end-index",
        type=int,
        default=None,
        help="End row index in manifest, exclusive.",
    )

    parser.add_argument(
        "--limit",
        type=int,
        default=None,
        help="Limit number of records. Useful for testing.",
    )

    parser.add_argument(
        "--stop-after-failures",
        type=int,
        default=0,
        help=(
            "If >0, stop a model after this many failures. "
            "Default 0 means continue through failures."
        ),
    )

    parser.add_argument(
        "--model-version",
        default=None,
        help=(
            "Optional exact checkpoint/API model version override. "
            "Use only when running one model at a time."
        ),
    )

    args = parser.parse_args()

    project_dir = Path(args.project_dir)
    ensure_project_dirs(project_dir)
    paths = project_paths(project_dir)

    models = parse_models(args.models)

    if args.model_version is not None and len(models) != 1:
        raise ValueError("--model-version can only be used when running one model.")

    manifest_df = read_manifest(project_dir)
    manifest_df = subset_manifest(
        manifest_df,
        record_id=args.record_id,
        start_index=args.start_index,
        end_index=args.end_index,
        limit=args.limit,
    )

    prompt = read_prompt(project_dir)

    if len(manifest_df) == 0:
        raise RuntimeError("No records selected for inference.")

    existing_keys = load_existing_result_keys(project_dir) if args.resume else set()

    print("=" * 100)
    print("AMR MEMBER 2 FULL INFERENCE")
    print("=" * 100)
    print("Project dir:", project_dir)
    print("Models:", models)
    print("Selected records:", len(manifest_df))
    print("Runs per record:", args.runs)
    print("Max new tokens:", args.max_new_tokens)
    print("Resume:", args.resume)
    print("Use 4-bit:", args.use_4bit)
    print("Results CSV:", paths["results_csv"])
    print("Raw outputs CSV:", paths["raw_outputs_csv"])
    print("Failures CSV:", paths["failures_csv"])
    print("=" * 100)

    for model_name in models:
        adapter = None
        model_failures = 0

        print("\n" + "#" * 100)
        print(f"STARTING MODEL: {model_name}")
        print("#" * 100)

        try:
            clear_gpu_memory()

            adapter = get_model_adapter(
                model_name=model_name,
                model_version=args.model_version,
                use_4bit=args.use_4bit,
            )

            print("Adapter model_name:", adapter.model_name)
            print("Adapter model_version:", adapter.model_version)
            print("Loading model...")

            adapter.load()

            print("Model loaded.")

            for _, record in manifest_df.iterrows():
                record_id = str(record["record_id"])
                image_path = Path(str(record["handwritten_image_path"]))
                ground_truth_text = str(record["ground_truth_text"])

                if not image_path.exists():
                    for run_num in range(1, args.runs + 1):
                        append_failure(
                            {
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "image_path": str(image_path),
                                "error_type": "FileNotFoundError",
                                "error_message": f"Image not found: {image_path}",
                                "created_at_utc": utc_now_iso(),
                            },
                            project_dir=project_dir,
                        )
                    continue

                if not ground_truth_text.strip():
                    for run_num in range(1, args.runs + 1):
                        append_failure(
                            {
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "image_path": str(image_path),
                                "error_type": "EmptyGroundTruth",
                                "error_message": "Ground truth text is empty; cannot score CER/WER.",
                                "created_at_utc": utc_now_iso(),
                            },
                            project_dir=project_dir,
                        )
                    continue

                for run_num in range(1, args.runs + 1):
                    result_key = (record_id, adapter.model_name, int(run_num))

                    if args.resume and result_key in existing_keys:
                        print(f"SKIP existing: {record_id} | {adapter.model_name} | run {run_num}")
                        continue

                    print(
                        f"\nRunning: record={record_id} | "
                        f"model={adapter.model_name} | run={run_num}"
                    )

                    try:
                        raw_output = adapter.generate(
                            image_path=image_path,
                            prompt=prompt,
                            max_new_tokens=args.max_new_tokens,
                        )

                        clean_output = remove_common_model_prefixes(raw_output)

                        # Save individual text file immediately.
                        text_output_path = save_text_output(
                            record_id=record_id,
                            model=adapter.model_name,
                            run=run_num,
                            text=raw_output,
                            project_dir=project_dir,
                        )

                        # Save raw output row immediately.
                        append_raw_output(
                            {
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "prompt": prompt,
                                "raw_output": raw_output,
                                "clean_output": clean_output,
                                "ground_truth_text": ground_truth_text,
                                "image_path": str(image_path),
                                "created_at_utc": utc_now_iso(),
                            },
                            project_dir=project_dir,
                        )

                        # Score immediately.
                        scored = score_single_output(
                            record_id=record_id,
                            model=adapter.model_name,
                            model_version=adapter.model_version,
                            run=run_num,
                            ground_truth_text=ground_truth_text,
                            raw_output=raw_output,
                            project_dir=project_dir,
                        )

                        append_result(scored, project_dir=project_dir)

                        # Update resume keys within same process.
                        existing_keys.add(result_key)

                        append_progress_log(
                            project_dir,
                            {
                                "created_at_utc": utc_now_iso(),
                                "status": "success",
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "cer": scored["cer"],
                                "wer": scored["wer"],
                                "output_path": str(text_output_path),
                            },
                        )

                        print(
                            f"SUCCESS: {record_id} | {adapter.model_name} | "
                            f"run {run_num} | CER={scored['cer']:.4f} | WER={scored['wer']:.4f}"
                        )

                    except Exception as e:
                        model_failures += 1
                        err = traceback.format_exc()

                        append_failure(
                            {
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "image_path": str(image_path),
                                "error_type": type(e).__name__,
                                "error_message": str(e),
                                "created_at_utc": utc_now_iso(),
                            },
                            project_dir=project_dir,
                        )

                        append_progress_log(
                            project_dir,
                            {
                                "created_at_utc": utc_now_iso(),
                                "status": "failed",
                                "record_id": record_id,
                                "model": adapter.model_name,
                                "model_version": adapter.model_version,
                                "run": run_num,
                                "cer": "",
                                "wer": "",
                                "output_path": "",
                            },
                        )

                        traceback_path = (
                            project_dir
                            / "logs"
                            / f"traceback_{adapter.model_name}_{record_id}_run{run_num}.txt"
                        )
                        traceback_path.write_text(err, encoding="utf-8")

                        print(
                            f"FAILED: {record_id} | {adapter.model_name} | run {run_num}"
                        )
                        print(f"Error: {type(e).__name__}: {e}")
                        print("Traceback saved to:", traceback_path)

                        if (
                            args.stop_after_failures > 0
                            and model_failures >= args.stop_after_failures
                        ):
                            raise RuntimeError(
                                f"Stopping {adapter.model_name} after "
                                f"{model_failures} failures."
                            )

        except Exception as outer_e:
            print("\n" + "!" * 100)
            print(f"MODEL-LEVEL FAILURE: {model_name}")
            print(type(outer_e).__name__, str(outer_e))
            print("Continuing to next model if any.")
            print("!" * 100)

        finally:
            if adapter is not None:
                print(f"Unloading model: {model_name}")
                adapter.unload()

            clear_gpu_memory()

    print("\n" + "=" * 100)
    print("FULL INFERENCE COMPLETE")
    print("=" * 100)
    print("Results CSV:", paths["results_csv"])
    print("Raw outputs CSV:", paths["raw_outputs_csv"])
    print("Failures CSV:", paths["failures_csv"])


if __name__ == "__main__":
    main()
````
